```python
import pandas as pd
```

```python
USD_data = pd.read_csv('/Users/yitik/Desktop/Dersler/For Python/InvestGuide/investment_ml/preprocessedData/USD_data.csv')
df = USD_data.copy()
df.info()
```

```python
df.drop(columns = 'Unnamed: 0', inplace = True)
df.head(1)
```

```python
X = df[['Yıl', 'Ay', 'Gun', 'Onceki_Gun_Fiyat', 'Haftalik_Ortalama']]
y = df['Alış (Forex)']
```

```python
from sklearn.linear_model import BayesianRidge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# En iyi parametrelerle optimize edilmiş pipeline
pipeline = make_pipeline(
    StandardScaler(),
    BayesianRidge(
        alpha_1=1e-05,   # GridSearch'ten gelen
        alpha_2=1e-07,   # GridSearch'ten gelen
        lambda_1=1e-07,  # GridSearch'ten gelen
        lambda_2=1e-05,  # GridSearch'ten gelen
        max_iter=300     #max iter
    )
)
```

```python
pipeline.fit(X, y)
```

```python
import joblib
```

```python
joblib.dump(pipeline, '/Users/yitik/Desktop/Dersler/For Python/InvestGuide/investment_ml/Models/USD_Model.pkl')
```
