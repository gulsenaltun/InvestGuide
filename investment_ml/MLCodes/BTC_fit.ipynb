```python
import pandas as pd
```

```python
XAU_data = pd.read_csv('/Users/yitik/Desktop/Dersler/For Python/InvestGuide/investment_ml/preprocessedData/XAU_data.csv')
df = XAU_data.copy()
```

```python
df.info()
```

```python
df.drop(columns = 'Unnamed: 0', inplace = True)
```

```python
X = df[['Acilis', 'Yil', 'Ay', 'Gun', 'HaftaninGunu', 'Dunku_Kapanis']]
y = df.Kapanis
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import HuberRegressor
```

```python
pipeline = make_pipeline(
    StandardScaler(),
    HuberRegressor(
        alpha=0.0001,  # GridSearch sonucu
        epsilon=1.1,   # GridSearch sonucu (Agresif outlier temizliği)
        max_iter=2000  # Garanti olsun diye iterasyonu artırdık
    )
)
```

```python
pipeline.fit(X, y)
```

```python
import joblib
```

```python
joblib.dump(pipeline, '/Users/yitik/Desktop/Dersler/For Python/InvestGuide/investment_ml/Models/BTC_Model.pkl')
```
